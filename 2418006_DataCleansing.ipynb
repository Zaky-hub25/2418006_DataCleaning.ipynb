CELL 4 — Membaca data mentah

```python
file_name = "TUGAS_DATA_SET_TRANSAKSI_CRYPTO_2418006.xlsx"

data_mentah = pd.read_excel(file_name, header=None)

print("Ukuran data mentah:", data_mentah.shape)

display(data_mentah.head(10))
```

CELL 5 — Menampilkan kondisi data mentah

```python
print("Data mentah:")
display(data_mentah)
```

CELL 6 — Menghapus baris kosong

```python
data_mentah = data_mentah.dropna(how='all').reset_index(drop=True)

print("Data setelah baris kosong dihapus:")
display(data_mentah.head())
```

CELL 7 — Menentukan header yang benar

```python
data_mentah.columns = data_mentah.iloc[0]

df = data_mentah.iloc[1:].copy()

df = df.reset_index(drop=True)

print("Nama kolom:")
print(df.columns.tolist())

display(df.head())
```

CELL 8 — Cek informasi dataset sebelum cleaning

```python
print("===== INFORMASI DATASET =====")

print("Jumlah baris  :", df.shape[0])
print("Jumlah kolom  :", df.shape[1])

print("\nTipe data:")
print(df.dtypes)
```

CELL 9 — Cek missing value

```python
print("===== CEK MISSING VALUE =====")

missing = df.isnull().sum()

display(missing.to_frame(name="Jumlah Missing"))
```

CELL 10 — Cek data duplikat

```python
print("===== CEK DATA DUPLIKAT =====")

jumlah_duplikat = df.duplicated().sum()

print("Jumlah data duplikat:", jumlah_duplikat)

if jumlah_duplikat > 0:
    display(df[df.duplicated(keep=False)])
else:
    print("Tidak terdapat data duplikat.")
```

CELL 11 — Membersihkan kolom teks

```python
kolom_teks = [
    'TX_ID',
    'Nama-Aset',
    'Jenis_Transaksi',
    'Status'
]

for kolom in kolom_teks:
    df[kolom] = df[kolom].astype(str).str.strip()

print("Data teks berhasil dibersihkan dari spasi berlebih.")
```

CELL 12 — Mengubah tipe data angka

```python
kolom_numerik = [
    'No',
    'Jumlah',
    'Harga_USDT'
]

for kolom in kolom_numerik:
    df[kolom] = pd.to_numeric(df[kolom], errors='coerce')

print("Kolom numerik berhasil dikonversi.")
```

CELL 13 — Cek tipe data setelah cleaning

```python
print("===== TIPE DATA SETELAH CLEANING =====")

print(df.dtypes)
```

CELL 14 — Menghapus duplikat

```python
jumlah_sebelum = len(df)

df = df.drop_duplicates().reset_index(drop=True)

jumlah_sesudah = len(df)

print("Jumlah data sebelum :", jumlah_sebelum)
print("Jumlah data sesudah :", jumlah_sesudah)
print("Duplikat dihapus    :", jumlah_sebelum - jumlah_sesudah)
```

CELL 15 — Validasi missing value

```python
print("===== VALIDASI MISSING VALUE =====")

missing_setelah = df.isnull().sum()

display(missing_setelah.to_frame(name="Jumlah Missing"))
```

CELL 16 — Cek nilai unik

```python
print("===== NILAI UNIK =====")

print("\nNama Aset:")
print(df['Nama-Aset'].unique())

print("\nJenis Transaksi:")
print(df['Jenis_Transaksi'].unique())

print("\nStatus:")
print(df['Status'].unique())
```

CELL 17 — Tampilkan dataset setelah cleaning

```python
print("===== DATASET SETELAH CLEANING =====")

display(df)
```

CELL 18 — Ringkasan hasil cleaning

```python
print("==============================================")
print("        HASIL DATA CLEANING")
print("==============================================")

print("Jumlah data akhir :", len(df))
print("Jumlah kolom      :", len(df.columns))
print("Jumlah duplikat   :", df.duplicated().sum())
print("Total missing     :", df.isnull().sum().sum())

print("\nKolom dataset:")
for kolom in df.columns:
    print("-", kolom)
```

CELL 19 — Simpan dataset bersih

```python
nama_file_bersih = "TUGAS_DATA_SET_TRANSAKSI_CRYPTO_2418006_CLEAN.xlsx"

df.to_excel(nama_file_bersih, index=False)

print("Dataset bersih berhasil disimpan:")
print(nama_file_bersih)
```
